```python
#count the ocuurence of each word in a line of text
text=input("Enter a  line of text:")
words=text.split()
count={}
for word in words:
    count[word]=count.get(word,0)+1
    print(count)
```

```python
# Count occurrences of 'a' within a list of first names
names = ["Arun", "Dev", "Das", "Ben","Adhil","Ajun"]
count = 0
for name in names:
    count = count + name.lower().count('a')
print("Number of a:", count)
```
